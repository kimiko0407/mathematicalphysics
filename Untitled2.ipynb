```python
import cmath
import math


z = 3 + 4j
print(f"Original Vector: {z} -> (Coordinates: {z.real}, {z.imag})")

translation_vector = 2 + 1j
z_translated = z + translation_vector
print(f"Translated:      {z_translated}")

scale_factor = 2.0
z_scaled = z * scale_factor
print(f"Scaled (x2):     {z_scaled}")


angle = math.pi / 2
rotation_multiplier = cmath.exp(1j * angle)
z_rotated = z * rotation_multiplier
print(f"Rotated 90°:     {z_rotated.real:.1f} + {z_rotated.imag:.1f}j")


z_reflected_x = z.conjugate()
print(f"Reflected (X):   {z_reflected_x}")

z_reflected_y = -z.conjugate()
print(f"Reflected (Y):   {z_reflected_y}")
```
